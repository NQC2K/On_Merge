```python
from databricks.sdk import WorkspaceClient

# Initialize client
w = WorkspaceClient()

# Commit information
prev_commit_id = "5c920ee37a2b171949c6fbde2240048f2b196f3b"
prev_commit_msg = "Update Synthetic_Data_CMDB.ipynb"
curr_commit_id = "5c920ee37a2b171949c6fbde2240048f2b196f3b"
curr_commit_msg = "Update Synthetic_Data_CMDB.ipynb"

# Job name
job_name = "synthetic_data_bank"
```

```python
# If commit IDs are the same, do nothing
if curr_commit_id == prev_commit_id:
    print("No changes detected. Job will not be triggered.")
else:
    print("Commit changed. Triggering Databricks job...")

    # Find the job by name
    job = next(
        (j for j in w.jobs.list() if j.settings.name == job_name),
        None
    )

    if job is None:
        raise ValueError(f"Job not found: {job_name}")

    print("Job ID:", job.job_id)

    # Trigger the job
    run = w.jobs.run_now(job_id=job.job_id)

    # Print run ID
    print("Run ID:", run.run_id)
```
